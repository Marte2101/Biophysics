```python
import numpy as np
import matplotlib.pyplot as plt

x, data = np.loadtxt('/home/C_potential',comments=["#", "@"], unpack = True)
x1, data1 = np.loadtxt('/home/A_potential',comments=["#", "@"], unpack = True)
x2, data2= np.loadtxt('/home/wt_potential.xvg',comments=["#", "@"], unpack = True)

y, datay = np.loadtxt('/home/C_pressure.xvg',comments=["#", "@"], unpack = True)
y1, datay1 = np.loadtxt('/home/A_pressure.xvg',comments=["#", "@"], unpack = True)
y2, datay2= np.loadtxt('/home/wt_pressure.xvg',comments=["#", "@"], unpack = True)

z, dataz = np.loadtxt('/home/C_temp.xvg',comments=["#", "@"], unpack = True)
z1, dataz1 = np.loadtxt('/home/A_temp.xvg',comments=["#", "@"], unpack  = True)
z2, dataz2= np.loadtxt('/home/wt_temperature.xvg', comments=["#","@"], unpack = True)


print("Average pressure C: ",  np.average(datay),  "Variance pressure C: ",  np.std(datay) )
print("Average pressure A: ", np.average(datay1), "Variance pressure A: ", np.std(datay1))
print("Average pressure WT: ", np.average(datay2), "Variance pressure WT: ", np.std(datay2))

print("Average temperature C: ", np.average(dataz), "Variance temperature C: ", np.std(dataz))
print("Average temperature A: ", np.average(dataz1), "Variance temperature A: ", np.std(dataz1))
print("Average temperature WT: ", np.average(dataz2), "Variance temperature WT: ", np.std(dataz))
```

```python
z, dataz = np.loadtxt('/home/C_temp.xvg',comments=["#", "@"], unpack = True)
z1, dataz1 = np.loadtxt('/home/A_temp.xvg',comments=["#", "@"], unpack  = True)
z2, dataz2= np.loadtxt('/home/wt_temperature.xvg', comments=["#","@"], unpack = True)

fig, axs = plt.subplots(nrows=1, ncols=3, figsize=(15, 5))

plt.subplot(1,3,1)
plt.plot(z, dataz)
plt.title("Chimera Mutant Temperature")
axs[0].set_xlabel('Time (ps)')
axs[0].set_ylabel('Temperature (K)')
axs[0].set_xlim(0, 110)
axs[0].set_ylim(290, 310)
plt.subplot(1,3,2)
plt.plot(z1, dataz1)
plt.title("Alpha Fold Mutant Temperature")
axs[1].set_xlabel('Time (ps)')
axs[1].set_ylabel('Temperature (K)')
axs[1].set_xlim(0, 110)
axs[1].set_ylim(290, 310)
plt.subplot(1,3,3)
plt.plot(z2, dataz2)
plt.title("Wild Type Temperature")
axs[2].set_xlabel('Time (ps)')
axs[2].set_ylabel('Temperature (K)')
axs[2].set_xlim(0, 110)
axs[2].set_ylim(290, 310)
```
